**Step 3: Implementing the Search Functionality**\
We'll create a function that allows users to input keywords related to issues, and searches the Known_Issues field for matches.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('mobile_products.csv')

def search_issues(keyword):
    # Convert keyword and Known_Issues to lowercase to make the search case-insensitive
    keyword_lower = keyword.lower()
    df['Known_Issues_Lower'] = df['Known_Issues'].str.lower()

    # Filter the DataFrame
    mask = df['Known_Issues_Lower'].str.contains(keyword_lower)
    matched_phones = df[mask]

    # Drop the temporary 'Known_Issues_Lower' column
    matched_phones = matched_phones.drop(columns=['Known_Issues_Lower'])
    return matched_phones

# User input
keyword = input("Enter issue keyword(s) to search for: ")

# Search for issues
results = search_issues(keyword)

# Display the results
if not results.empty:
    print("\nMobile phones with issues matching your search:\n")
    print(results[['Model_Name', 'Known_Issues']])
else:
    print("\nNo mobile phones found with issues matching your search.")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('mobile_products.csv')

def search_issues(keywords):
    keyword_list = keywords.lower().split()
    pattern = '|'.join(keyword_list)
    mask = df['Known_Issues'].str.lower().str.contains(pattern)
    matched_phones = df[mask]
    return matched_phones

def main():
    while True:
        keyword = input("Enter issue keyword(s) to search for (or type 'exit' to quit): ")
        if keyword.lower() == 'exit':
            print("Exiting the program.")
            break
        results = search_issues(keyword)
        if results.empty:
            print("\nNo mobile phones found with issues matching your search.\n")
            continue

        # Optional: Filter by price
        filter_price = input("Do you want to filter by price? (yes/no): ")
        if filter_price.lower() == 'yes':
            try:
                min_price = float(input("Enter minimum price: "))
                max_price = float(input("Enter maximum price: "))
                results = results[(results['Price'] >= min_price) & (results['Price'] <= max_price)]
                if results.empty:
                    print("\nNo mobile phones found within the specified price range.\n")
                    continue
            except ValueError:
                print("Invalid price entered. Skipping price filtering.")

        # Display results
        print("\nMobile phones with issues matching your search:\n")
        for idx, row in results.iterrows():
            print(f"Model Name: {row['Model_Name']}")
            print(f"Features: {row['Features']}")
            print(f"Release Date: {row['Release_Date']}")
            print(f"Price: ${row['Price']}")
            print(f"Known Issues: {row['Known_Issues']}\n")

if __name__ == "__main__":
    main()
```
